```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
data['scaled'] = scaler.fit_transform(data[['AEP_MW']])
```

```python
import numpy as np

def create_sequence(data, look_back):
  X = []   # X will hold sequence of inputs
  y = []   # y will hold the corresponding target values(i.e, the next value to predit)
  for i in range(look_back, len(data)):
    X.append(data[i - look_back:i])
    y.append(data[i])
  return np.array(X), np.array(y)
```

```python
look_back = 24
X, y = create_sequence(data['scaled'], look_back)
print("X shape:", X.shape)
print("Y shape:", y.shape)
```

```python
import matplotlib.pyplot as plt
plt.plot(X[0], label="Input Sequence")
plt.axhline(y=y[0], color='red', linestyle='--', label="Target")
plt.title("Sample Input Sequence and Target")
plt.legend()
plt.grid(True)
plt.show()
```

```python
for look_back in [24, 48, 72]:
  X,y = create_sequence(data['scaled'], look_back)

  plt.figure(figsize=(10, 4))
  plt.plot(X[0], label = f"Input Sequence (Look Back={look_back})")
  plt.axhline(y[0], color='red', linestyle='--', label='Target')
  plt.title(f'Input vs Target for Look Back={look_back}')
  plt.xlabel('Time Steps')
  plt.ylabel('Normalized Consumption')
  plt.legend()
  plt.grid(True)
  plt.show()
```

```python
import numpy as np
from matplotlib import pyplot as plt

look_back = 48
X, y = create_sequence(data['scaled'], look_back)
train_size = int(len(X) * 0.9)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

plt.figure(figsize=(14, 5))
plt.plot(y, label='Full Data', color='lightgray')
plt.axvline(x = train_size, color ='red', linestyle='--', label='Train/Test Split')
plt.plot(range(train_size), y[:train_size], label='Train Targets', color = 'blue')
plt.plot(range(train_size, len(y)), y[train_size:], label='Test Targets', color='green')

plt.title('Train/Test Split on Target Data')
plt.xlabel('Time Step')
plt.ylabel('Normalized Consumption')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

print(X.shape)
X = X.reshape((X.shape[0], X.shape[1], 1))
train_size = int(len(X) * 0.9)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]
print(X_train.shape)


model = Sequential()
model.add(LSTM(64, return_sequences=True, input_shape=(X_train.shape[1], X_train.shape[2])))
model.add(Dropout(0.2)) # 20% of the outputs will be randomly dropped during training

model.add(LSTM(32))
model.add(Dropout(0.2))

model.add(Dense(1))
model.compile(loss='mse', optimizer='adam')
model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))
X_test = X_test.reshape((X_test.shape[0], X_test.shape[1], 1))

#1. Defining Callbacks
early_stop = EarlyStopping(
    monitor = 'val_loss',
    patience = 10,
    restore_best_weights = True
)

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    verbose=1
)

#2. Fit model
history = model.fit(
    X_train, y_train,
    epochs=20,
    batch_size=32,
    validation_data=(X_test, y_test),
    callbacks=[early_stop, reduce_lr],
    verbose=1
)
```

```python
print(X_train.shape)
print(y_train.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Mean Squared Error (MSE)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error

# Make predictions
y_pred = model.predict(X_test)

# Calculate evaluation metrics
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print(f"Mean Squared Error (MSE): {mse:.6f}")
print(f"Mean Absolute Error (MAE): {mae:.6f}")
```
